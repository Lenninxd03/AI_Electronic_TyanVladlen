```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# Length and height data for Dachshunds
dach_length = [55, 57, 64, 63, 58, 49, 54, 61]
dach_height = [30, 31, 36, 30, 33, 25, 37, 34]

# Length and height data for Jindo dogs
jin_length = [56, 47, 56, 46, 49, 53, 52, 48]
jin_height = [52, 52, 50, 53, 50, 53, 49, 54]

# Create labels: Dachshund = 0, Jindo = 1
X = np.array(
    [[l, h] for l, h in zip(dach_length + jin_length,
                          dach_height + jin_height)]
)
y = np.array([0]*len(dach_length) + [1]*len(jin_length))

# Train logistic regression model
model = LogisticRegression()
model.fit(X, y)

# New data for prediction
newdata_length = [59]
newdata_height = [35]
X_new = np.array([[newdata_length[0], newdata_height[0]]])

# Make prediction and get probabilities
y_pred = model.predict(X_new)[0]
y_prob = model.predict_proba(X_new)[0]

breed_names = {0: 'Dachshund', 1: 'Jindo dog'}
print(f"\n--- Prediction for New Dog (Length: {newdata_length[0]}, Height: {newdata_height[0]}) ---")
print(f"Predicted Breed: {breed_names[y_pred]}")
print(f"Probabilities -> Dachshund: {y_prob[0]:.4f}, Jindo: {y_prob[1]:.4f}")

# ---- Visualization ----
plt.figure(figsize=(8, 6))
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')

# Plot the decision boundary
x_values = np.linspace(min(X[:,0])-2, max(X[:,0])+2, 200)
w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b  = model.intercept_[0]
y_values = -(w1 * x_values + b) / w2
plt.plot(x_values, y_values, 'g--', linewidth=2, label='Decision Boundary')

# Plot the new data point
plt.scatter(newdata_length, newdata_height, c='purple', marker='X', s=150, label='New dog found by TYAN VLADLEN')

plt.xlabel('Length')
plt.ylabel('Height')
plt.title("Dog Classification with Decision Line & New Data")
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```
