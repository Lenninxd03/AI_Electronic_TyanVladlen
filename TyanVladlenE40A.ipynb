```python
# TYAN VLADLEN E40A

import matplotlib.pyplot as plt
import numpy as np

x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

mymodel = np.poly1d(np.polyfit(x, y, 2))

a, b, c = mymodel.coefficients


def f(x_value):
    return mymodel(x_value)


print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

time_to_predict = 17
predicted_speed = f(time_to_predict)

print(f"Predicted speed at 17:00 (5:00 PM): {predicted_speed:.2f}")

myline = np.linspace(1, 22, 100)

plt.scatter(x, y, color="blue", label="Actual Data")
plt.plot(myline, f(myline), color="red", label="Regression Line")

plt.scatter(
    time_to_predict,
    predicted_speed,
    color="green",
    s=100,
    zorder=5,
    label=f"Predicted at 17:00 ({predicted_speed:.2f})",
)
plt.annotate(
    f"({time_to_predict}, {predicted_speed:.2f})",
    xy=(time_to_predict, predicted_speed),
    xytext=(time_to_predict - 3, predicted_speed + 5),
    arrowprops=dict(facecolor="black", shrink=0.05, width=1, headwidth=6),
)

plt.title("Tyan Vladlen E40A, Polynomial Regression (degree=2)")
plt.xlabel("Time (Hour)")
plt.ylabel("Speed")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
#######
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract the coefficients in the form f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Define the prediction function
def f(x_value):
    return mymodel(x_value)

# Print the prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values for prediction
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```
